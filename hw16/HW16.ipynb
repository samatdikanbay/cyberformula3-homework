### Задача 2. Дорогие заказы Шымкента
Добавьте столбец `revenue` = `price` × `qty`. Выведите 5 самых дорогих **оплаченных** заказов из Шымкента: только столбцы `date`, `product`, `qty`, `revenue`.

```python
# Задача 2
#Без ИИ все сложней справляться
df["revenue"] = df["price"] * df["qty"]
shymkent_paid = df[(df["city"] == "Шымкент") & (df["status"] == "оплачен")]
top_5 = shymkent_paid.sort_values(by="revenue", ascending=False).head(5)[["date", "product", "qty", "revenue"]]
print(top_5)
```

### Задача 3. Покупают помногу?
1. Сколько заказов, в которых купили **2 штуки или больше**? Какая это доля всех заказов в %?
2. Какой товар чаще всего покупают сразу **по 3 штуки**?

```python
# Задача 3
#1
orders_2_plus = df[df["qty"] >= 2]
count_2_plus = len(orders_2_plus)
share_2_plus = (count_2_plus / len(df)) * 100

print(f"Заказов больше 2 шт.: {count_2_plus}")
print(f"Доля от всех заказов: {share_2_plus:.2f}%")
```

```python
#2
orders_3 = df[df["qty"] == 3]
top_product_3 = orders_3["product"].value_counts().idxmax()

print("Чаще всего по 3 шт. покупают: {top_product_3}")
```
